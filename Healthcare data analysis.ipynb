```python
df["Length of Stay"] =(
    pd.to_datetime(df["Discharge Date"]) -
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission","Discharge Date","Length of Stay"]].head()
```

```python
df["Admission Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.month

df[["Date of Admission","Admission Month"]].head()
```

```python
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
monthly_admissions =(

df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)

monthly_admissions
```

```python
plt.figure(figsize=(9,5))

sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Name",
    marker="o"
)
plt.title("Monthly Number of admissions")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Name",
   
)
plt.title("Monthly Number of admissions")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()
```

```python
plt.figure(figsize=(9,5))

monthly_admissions["Admission Month"].value_counts().sort_index().plot(kind="bar")

plt.title("Number of Admissions by Month")
plt.xlabel("Admission Month")
plt.ylabel("Number of Admissions")

plt.show()
```

```python
plt.figure(figsize=(8,8))

monthly_admissions["Admission Month"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Monthly Admission Distribution")
plt.ylabel("")

plt.show()
```
