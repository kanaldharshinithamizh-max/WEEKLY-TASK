```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("SHOP_2015-05-21_2025-03-16.csv.xls")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)

df["date"] = pd.to_datetime(df["date"], utc=True)

df.isnull().sum()
```

```python
df.select_dtypes(include="number").corr()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["date"], df["close"])
plt.title("Closing Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["date"], df["volume"])
plt.title("Trading Volume")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["close"], bins=30)
plt.title("Closing Price Distribution")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["close"])
plt.title("Closing Price Boxplot")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

# EXPLORATORY DATA ANALYSIS (EDA) REPORT
# Dataset Name: Shopify (SHOP) Historical Stock Data
# 1. Introduction
Exploratory Data Analysis (EDA) is the process of analysing a dataset to understand its structure, identify patterns and find errors. In this project, EDA is performed on Shopify stock market data using Python libraries.
# 2. Objectives
- To understand the dataset.
- To identify missing values.
- To check duplicate records.
- To perform data cleaning.
- To calculate statistical information.
- To analyse relationships between variables.
- To visualize data using graphs.
# 3. Dataset Description
The Shopify dataset contains historical stock market information from 2015 to 2025.

The dataset contains 2,469 rows and 7 columns.

The columns are:
1. Date – Date of stock trading.
2. Open – Opening stock price.
3. High – Highest stock price.
4. Low – Lowest stock price.
5. Close – Closing stock price.
6. Adj Close – Adjusted closing price.
7. Volume – Number of shares traded.
8. Data Inspection
Data inspection is used to understand the dataset.

# The following functions are used:
- head() – Displays the first five rows.
- tail() – Displays the last five rows.
- shape – Displays the number of rows and columns.
- info() – Displays column names and data types.
- describe() – Displays statistical information.
# 5. Data Cleaning
Data cleaning is performed to improve the quality of the dataset.

The following operations are performed:
- Checking missing values.
- Checking duplicate values.
- Removing duplicate records.
- Converting the date column into datetime format.
- Sorting records based on date.
# Result: The dataset contains no missing values and no duplicate records.
# 6. Descriptive Statistics
Descriptive statistics help us understand numerical data.

The describe() function is used to calculate:
- Count
- Mean
- Standard Deviation
- Minimum
- Maximum
- 25%, 50% and 75% quartiles
These values help us understand the stock price and trading volume.
# 7. Correlation Analysis
Correlation analysis is used to identify relationships between numerical variables.

The corr() function is used to calculate correlation values.

Correlation values range from -1 to +1.
- +1 indicates a positive relationship.
- -1 indicates a negative relationship.
- 0 indicates no linear relationship.
A heatmap is used to display the correlation between variables.
# 8. Data Visualization
Data visualization helps us understand the dataset using graphs.

# The following graphs are created:

1. Closing Price Graph

It shows the changes in Shopify stock closing prices over time.

2. Trading Volume Graph

It displays the number of shares traded over time.
3. Histogram

It shows the distribution and frequency of closing prices.

4. Boxplot

It displays the median, spread and possible outliers in closing prices.

5. Correlation Heatmap

It shows the relationships between numerical columns using different colours.
# 9. Conclusion
Exploratory Data Analysis was successfully performed on the Shopify stock dataset using Python. The dataset was inspected, cleaned and analysed using statistical methods. Different visualizations were created to understand stock prices, trading volume and relationships between variables.

Thus, EDA helps us understand the dataset and identify useful patterns from historical stock market data.
